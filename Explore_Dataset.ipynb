```python
from pandasql import sqldf

# Global query function
pysqldf = lambda q: sqldf(q, globals())
```

```python
sql = """
SELECT *
FROM (
    SELECT *
    FROM df
    WHERE mentioned = 'Non-Tobacco Company'
    LIMIT 50
)

UNION ALL

SELECT *
FROM (
    SELECT *
    FROM df
    WHERE mentioned != 'Non-Tobacco Company'
       OR mentioned IS NULL
    LIMIT 50
)
"""
```

```python
df_mentioned = pysqldf(sql)
```

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
analyzer = SentimentIntensityAnalyzer()
```

```python
def vader_first_500(text):
    if not isinstance(text, str) or not text.strip():
        return None
    short_text = text[:500]
    return analyzer.polarity_scores(short_text)['compound']

df['vader_compound_500'] = df['ocr_text'].apply(vader_first_500)
```

```python
df
```

```python
sql_sentiment = """
SELECT
  CASE
    WHEN mentioned = 'Non-Tobacco Company' THEN 'Non-Tobacco Company'
    ELSE 'Other'
  END AS mentioned_group,
  AVG(vader_compound_500)                    AS avg_sentiment,
  SUM(CASE WHEN vader_compound_500 IS NOT NULL THEN 1 ELSE 0 END) AS n_non_null,
  COUNT(*)                                   AS total_rows
FROM df
GROUP BY mentioned_group;
"""
```

```python
pysqldf(sql_sentiment)
```
